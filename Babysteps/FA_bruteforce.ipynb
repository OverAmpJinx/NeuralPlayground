# Brute Force Variant of the Function Approximation Example

Here i just implement the neuron, loss function and optimizer by hand instead of relying on the nn from torch lib

```python
import sys
import torch
import matplotlib.pyplot as plt
import numpy as np
```

```python
device = torch.device("mps" if torch.backends.mps.is_available() else "cpu")
```

```python
x_data = torch.linspace(2, 20, 400, device=device).unsqueeze(1)
y_data = torch.sin(3*x_data) * torch.exp(-0.4 * x_data)
```

```python
# defining functions of model, lossfunc and optimizer
class Neuron:
    def __init__(self, n_inputs):
        self.w =np.random.randn(n_inputs)
        self.b = 0
    def forward(self, x):
        return np.dot(x, self.w) + self.b

def lossfunc(y_pred, y_true):
    return np.sum((y_pred - y_true) ** 2)

y_true = np.array([1.0, 2.0, 3.0])
y_pred = np.array([1.5, 2.0, 2.5])

loss = lossfunc(y_pred, y_true)
print(loss)
```
